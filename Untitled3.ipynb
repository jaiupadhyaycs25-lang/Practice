Write a file using ; sd the delimiter and -999 for missing values.Read so -999 becomes Nan

```python
import numpy as np
import pandas as pd

# 1. Create sample data with missing values represented by -999, separated by semicolons
data = """101;Alice;Engineering;NY;85000
102;Bob;-999;SF;92000
103;Charlie;Sales;-999;78000"""

with open("sample_data_semicolon.csv", "w") as f:
    f.write(data)


df = pd.read_csv(
    "sample_data_semicolon.csv",
    sep=";",
    header=None,
    names=["ID", "Name", "Department", "City", "Salary"],
    na_values=[-999],
    index_col=0,
)

# Display the resulting DataFrame
print(df)
```

Read a big csv in chunks of 2000 rows and compute the mean of one column per key across all the chunks

```python
import zipfile
import pandas as pd

# 1. Zip file ko extract karein
with zipfile.ZipFile("archive (6).zip", "r") as zip_ref:
  zip_ref.extractall("extracted_dataset")

# 2. Sahi file path dein (Titanic dataset ka naam yahan use kiya hai)
file_path = "extracted_dataset/Titanic-Dataset.csv"
chunk_size = 2000

# 3. Titanic dataset ke columns ke mutabiq Grouping aur Target set karein

key_column = "Pclass"
target_column = "Age"

total_sum = pd.Series(dtype=float)
total_count = pd.Series(dtype=float)

# 4. Chunk-by-chunk read karke sum aur count calculate karein
for chunk in pd.read_csv(file_path, chunksize=chunk_size):
  grouped = chunk.groupby(key_column)[target_column]

  chunk_sum = grouped.sum()
  chunk_count = grouped.count()

  total_sum = total_sum.add(chunk_sum, fill_value=0)
  total_count = total_count.add(chunk_count, fill_value=0)

# 5. Final combined mean nikalen
overall_mean = total_sum / total_count

print("\nOverall Mean per Key:")
print(overall_mean)
```

## Export the Dataframe with orient = 'records",read it back with pd.read_json, and check equality with .equals

```python
df.to_json("dataframe_records.json", orient="records")

df_reloaded = pd.read_json("dataframe_records.json")

are_equal = df.reset_index().equals(df_reloaded.reset_index())

print("Are the DataFrames equal?", are_equal)
```
